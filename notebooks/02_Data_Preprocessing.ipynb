##  Check Missing Values

```python
df.isnull().sum()
```

##  Check Duplicate Records

```python
df.duplicated().sum()
```

##  Check Unique Values

```python
print("Gender values:")
print(df["Gender"].unique())

print("\nGender counts:")
print(df["Gender"].value_counts())
```

##  Check Invalid/Unusual Values

```python
print("Age range:", df["Age"].min(), "to", df["Age"].max())

print("Income range:",
      df["Annual Income (k$)"].min(),
      "to",
      df["Annual Income (k$)"].max())

print("Spending Score range:",
      df["Spending Score (1-100)"].min(),
      "to",
      df["Spending Score (1-100)"].max())
```

```python
processed_df = df.copy()
```

```python
processed_df.to_csv(
    "../data/processed/processed_mall_customers.csv",
    index=False
)
```

## Preprocessing Summary

The Mall Customers dataset was inspected and prepared for further analysis. Missing values, duplicate records, data types, categorical values, and numerical ranges were checked. CustomerID was excluded from clustering features because it is an identifier rather than a meaningful customer attribute. The numerical customer attributes Age, Annual Income, and Spending Score were selected for further analysis. The processed dataset is now ready for exploratory data analysis and feature preparation.
