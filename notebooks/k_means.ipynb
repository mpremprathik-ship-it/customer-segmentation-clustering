# K-Means Clustering

## Objective

To segment customers into 5 clusters using the K-Means clustering algorithm based on scaled Annual Income and Spending Score.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
X_scaled = pd.read_csv("../data/scaled_customer_data.csv")
```

```python
X_scaled
```

```python
X_scaled.head()
```

```python
X_scaled.shape
```

```python
X_scaled.columns
```

```python
X = X_scaled.values
```

```python
X[:5]
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)
```

```python
kmeans.fit(X)
```

```python
labels = kmeans.labels_
```

```python
labels[:10]
```

```python
df = pd.read_csv("../data/raw/Mall_Customers.csv")
```

```python
df.head()
```

```python
df["Cluster"] = labels
```

```python
df.head()
```

```python
df["Cluster"].value_counts().sort_index()
```

```python
df.to_csv(
    "../data/customer_segments.csv",
    index=False
)
```
